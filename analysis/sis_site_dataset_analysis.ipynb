```python
import json
import pandas as pd
from collections import Counter

with open("cc-site-info.json", "r") as f:
    cc_data = json.load(f)

# Filter out CCAI courses as agreed
cc_filtered = {k: v for k, v in cc_data.items() if not k.strip().upper().startswith("CCAI")}
print(f"Total active non-CCAI courses in cc-site-info.json: {len(cc_filtered)}\n")

study_load_keys = Counter()
ass_dist_keys = Counter()
non_numeric_loads = []
non_numeric_ass = []

for code, info in cc_filtered.items():
    # 1. Inspect study_load_dist
    for k, v in info.get("study_load_dist", {}).items():
        clean_k = k.strip()
        study_load_keys[clean_k] += 1
        # Check if value is a clean number
        if not str(v).strip().replace(".", "", 1).isdigit():
            non_numeric_loads.append((code, clean_k, v))

    # 2. Inspect ass_dist
    for k, v in info.get("ass_dist", {}).items():
        clean_k = k.strip()
        ass_dist_keys[clean_k] += 1
        # Check if value is a clean number (strip % first)
        v_clean = str(v).strip().replace("%", "").strip()
        if not v_clean.replace(".", "", 1).isdigit():
            non_numeric_ass.append((code, clean_k, v))

print(f"Unique 'study_load_dist' activity names: {len(study_load_keys)}")
print("Top 20 Study Load Keys:")
for k, count in study_load_keys.most_common(20):
    print(f"  - {k!r}: {count} courses")

print("\n" + "="*50 + "\n")

print(f"Unique 'ass_dist' assessment task names: {len(ass_dist_keys)}")
print("Top 30 Assessment Task Keys:")
for k, count in ass_dist_keys.most_common(30):
    print(f"  - {k!r}: {count} courses")

print("\n" + "="*50 + "\n")
print(f"Non-numeric study_load values found: {len(non_numeric_loads)}", non_numeric_loads[:10])
print(f"Non-numeric ass_dist values found: {len(non_numeric_ass)}", non_numeric_ass[:10])
```
